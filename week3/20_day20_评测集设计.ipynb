## 2. 打分标准（明天跑评测时用，今天先定好）

客观题打两项分，每题满分 10 分：

| 评分项 | 分值 | 标准 |
|---|---|---|
| **正确性** | 6 分 | 选对 6 分；选错 0 分 |
| **依据质量** | 4 分 | 给了正确出处/依据 4 分；依据错误或编造 0 分；没给 2 分 |

> 依据质量是**你独有的设计**：通用评测只判对错，你额外要求模型"给依据"——这样能同时测出 D17 发现的**幻觉**问题。面试讲这个设计，非常加分。
