## 1. 字符串索引与切片（复习）

字符串是「字符的列表」，支持下标和切片。

````python
s = "高血压 2型糖尿病"

print(s[0])       # 第 1 个字符
print(s[-1])      # 最后 1 个字符
print(s[0:3])     # 前 3 个字符（不含下标 3）
print(len(s))     # 长度
print("糖尿病" in s)   # 是否包含 → True
````

## 2. 去空白、大小写

````python
raw = "   WBC 4.2  "
print(repr(raw))             # repr 能看到两头的空格
print(repr(raw.strip()))     # strip() 去掉首尾空白
print(raw)

lab = "wbc"
print(lab.upper())           # 大写 → WBC
print("HGB".lower())         # 小写 → hgb
print("2型糖尿病".replace("型", "形 "))   # 替换
````

## 3. 查找与判断

````python
note = "患者主诉：发热、咳嗽三天"

print("发热" in note)                    # 是否包含 → True
print(note.find("咳嗽"))                 # 首次出现的位置（找不到返回 -1）
print(note.startswith("患者"))           # 是否以…开头 → True
print(note.endswith("三天"))             # 是否以…结尾 → True
````

## 4. 拆分与拼接（今天最重要）

`split()` 按分隔符切开成列表；`join()` 把列表拼回字符串。

````python
labs = "WBC,HGB,PLT,CRP"
items = labs.split(",")          # 按逗号拆成列表
print(items)                       # ['WBC', 'HGB', 'PLT', 'CRP']
print(items[1])                    # HGB

print(" | ".join(items))           # 用 ' | ' 拼回字符串

# split 还能拆「键值对」
field = "姓名:王小明"
key, value = field.split(":")     # 拆成两个
print(key, "=", value)
````

## 5. 医学实战：解析一条结构化病历

医院导出的数据常是 `姓名:王小明|年龄:45|诊断:2型糖尿病` 这种格式，拆成字典：

````python
record = "姓名:王小明|年龄:45|诊断:2型糖尿病"

patient = {}                     # 空字典
for part in record.split("|"):   # 先按 | 拆成 3 段
    key, value = part.split(":")
    patient[key] = value.strip()

print(patient)
print(patient["诊断"])            # 2型糖尿病
````

提示：先 split，再用循环或列表推导式 `.lower()`，最后 join。

**练习 5（今天的检查点·病历解析器）**：解析这条病历，存成字典并打印每个字段：

```python
record = "姓名:李小红|性别:女|年龄:32|诊断:社区获得性肺炎|过敏史:无"
```

要求：① 先按 `|` 拆；② 每段按 `:` 拆成键和值；③ 存入字典；④ 打印整个字典。

````python
# 在这里写你的代码
patient={}
record = "姓名:李小红|性别:女|年龄:32|诊断:社区获得性肺炎|过敏史:无"
for e in record.split("|"):
    key,value=e.split(":")
    patient[key] = value.strip()
print(patient)
````

**练习 6（加分题）**：写函数 `parse_record(text)`，把任意 `键:值|键:值|...` 格式的文本解析成字典并返回。然后测试练习 5 那条数据。

````python
# 在这里写你的代码
def parse_record(text):
    patient = {}
    for part in text.split("|"):
        key, value = part.split(":")
        patient[key] = value.strip()
    return patient

print(parse_record("姓名:李小红|性别:女|年龄:32|诊断:社区获得性肺炎|过敏史:无"))
````
