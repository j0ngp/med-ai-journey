## 1. groupby 是什么

`groupby("某列")` = 「按某列分组」。分组后就能对**每一组**分别算统计。

最经典的写法（分三步读）：

```python
df.groupby("诊断")["年龄"].mean()
```

1. `df.groupby("诊断")`：按「诊断」分组（糖尿病一组、高血压一组…）；
2. `["年龄"]`：每组里，只看「年龄」这一列；
3. `.mean()`：每组的年龄求平均。

````python
df.groupby("诊断")["年龄"].mean()    # 每种诊断的平均年龄
````

## 2. 常用的几种聚合

`mean`（均值）、`count`（人数）、`max`、`min`、`median`（中位数）、`sum`。

````python
df.groupby("诊断")["年龄"].count()     # 每种诊断各有多少人
````

````python
df.groupby("性别")["收缩压"].mean()    # 男女平均收缩压
````

````python
df.groupby("诊断")["血糖"].max()      # 每种诊断里血糖的最高值
````

## 3. agg：一次算多个统计

想同时看「每组的人数 + 平均年龄」，用 `agg` 传一个列表：

````python
df.groupby("诊断")["年龄"].agg(["count", "mean"])
````

````python
df.groupby("诊断")["收缩压"].agg(["mean", "max", "min"])
````
