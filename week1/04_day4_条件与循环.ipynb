## 1. if —— 让程序做判断

格式（注意冒号和缩进）：

```
if 条件:
    条件成立时执行的代码
elif 另一个条件:
    上面不成立、这个成立时执行
else:
    都不成立时执行
```

````python
temp = 38.5

if temp >= 39.0:
    print("高热")
elif temp >= 37.3:
    print("发热")
else:
    print("体温正常")

print("判断完毕")   # 不在 if 里，缩进不同，永远执行
````

**比较运算符**：`>` 大于 ｜ `<` 小于 ｜ `>=` 大于等于 ｜ `<=` 小于等于 ｜ `==` 等于 ｜ `!=` 不等于

⚠️ 判断相等用**两个等号** `==`；一个等号 `=` 是赋值。

### 1.1 输入 + 判断（组合技）

````python
glucose = float(input("输入血糖值（mmol/L）："))

if glucose < 3.9:
    print("低血糖")
elif glucose <= 6.1:
    print("正常")
else:
    print("偏高")
````

## 2. for —— 把一件事重复 N 次

两种用法：**遍历列表**、**配合 range 数数**。

````python
patients = ["王小明", "李小红", "张三"]

for name in patients:          # 每次循环，name 依次拿到列表里的一个元素
    print("患者：", name)
````

````python
for i in range(1, 6):          # range(1,6) = 1,2,3,4,5（不含 6）
    print("第", i, "号床位")
````

### 2.1 for + if 组合：遍历血糖值，逐个判断

````python
glucose_list = [5.2, 6.1, 4.8, 7.0, 3.5]

for g in glucose_list:
    if g < 3.9:
        print(g, "→ 低血糖")
    elif g <= 6.1:
        print(g, "→ 正常")
    else:
        print(g, "→ 偏高")
````

## 3. while —— 条件成立就一直重复

⚠️ 一定要有「停下来」的条件，否则会**死循环**（卡死时按 ⏹ 停止按钮）。

````python
count = 1
while count <= 3:        # 条件成立才继续
    print("测体温，第", count, "次")
    count = count + 1    # 每次 +1，不然永远停不下来

print("测量完毕")
````

### 3.1 猜数字（while 的经典用法）

````python
secret = 7
guess = int(input("猜猜患者的住院天数（1-10）："))

while guess != secret:                 # 没猜对就一直循环
    if guess > secret:
        print("猜大了")
    else:
        print("猜小了")
    guess = int(input("再猜一次："))    # 更新，否则死循环

print("猜对了！")
````
