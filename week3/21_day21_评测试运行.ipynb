## 第 3 步 · 用 pandas 打分（练习）

把你的记录写成一个 DataFrame（替换下面的示例数据），按打分标准计算：

- **正确性**：选对 6 分，选错 0 分；
- **依据**：依据正确 4 分；无依据 2 分；依据错误/编造 0 分。

```python
# 示例：前 3 题的记录（用你的真实记录替换）
records = pd.DataFrame({
    "pilot题号": [1, 2, 3],
    "标准答案":   ["B", "A", "D"],       # ← 从 pilot 表里抄
    "模型答案":   ["B", "C", "D"],
    "依据情况":   ["正确", "错误", "无依据"],   # 正确/无依据/错误
})
records
```

```python
# 打分
records["正确性得分"] = (records["标准答案"] == records["模型答案"]) * 6
records["依据得分"] = records["依据情况"].map({"正确": 4, "无依据": 2, "错误": 0})
records["总分"] = records["正确性得分"] + records["依据得分"]
records
print("平均分:", records["总分"].mean())
print("准确率:", (records["正确性得分"] > 0).mean())
```
