## Part B · CBLUE 入门：5 分钟搞清楚它是什么

**CBLUE**（Chinese Biomedical Language Understanding Evaluation）＝ 中文医疗 AI 的「执医考试」：
- 8 个子任务、5 大类：**信息抽取**（CMeEE 实体识别 / CMeIE 关系抽取）、**术语归一化**（CHIP-CDN）、**文本分类**（CHIP-CTC 临床试验筛选 / KUAKE-QIC 查询意图）、**句子关系判定**（CHIP-STS / KUAKE-QQR）、**问答**（KUAKE-QTR）；
- 官方仓库：[CBLUEbenchmark/CBLUE](https://github.com/CBLUEbenchmark/CBLUE)，数据可下载；
- **PromptCBLUE**（CCKS2023）：给大模型时代做的升级版——94 套提示模板、few-shot 评测 18 个子任务，与你的 D23 评测技能同族（原天池赛已结束，但数据开源：[michael-wzhu/PromptCBLUE](https://github.com/michael-wzhu/PromptCBLUE)）。

**诚实说明**：CBLUE 经典榜是静态的、2023 的天池赛已闭榜——我们的目标是**完整跑通一个子任务的 baseline + 官方指标**，把流程和分数写进简历/GitHub（含金量在于「跑通了」，不在于刷榜）。想打活跃榜单的话，盯 2026 年 **CCKS/CHIP 评测**（每年 10–11 月集中发题），到时报名即可。
