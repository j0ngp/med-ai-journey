## 1. 写文件：open + write + close

三种模式记牢：`"w"` 写入（覆盖）、`"a"` 追加、`"r"` 读取。

````python
f = open("note.txt", "w", encoding="utf-8")   # w 模式：新建/覆盖
f.write("患者：王小明\n")                       # \n 是换行
f.write("诊断：2型糖尿病\n")
f.close()                                       # 用完必须关

print("已写入 note.txt")
````

### 1.1 推荐写法：with open（自动关闭，不怕忘 close）

````python
with open("note2.txt", "w", encoding="utf-8") as f:
    f.write("自动关闭的写法\n")

# 缩进块结束，文件已自动关闭
print("已写入 note2.txt")
````

## 2. 读文件

````python
with open("note.txt", "r", encoding="utf-8") as f:
    content = f.read()          # 一次性读全部
print(content)
````

````python
with open("note.txt", "r", encoding="utf-8") as f:
    for line in f:              # 逐行读（处理大文件用这个）
        print(line.strip())     # strip 去掉行尾换行符
````

## 3. 追加："a" 模式（不覆盖，往后加）

````python
with open("note.txt", "a", encoding="utf-8") as f:
    f.write("随访：3个月后复诊\n")

with open("note.txt", "r", encoding="utf-8") as f:
    print(f.read())
````

## 4. CSV 文件：用 csv 模块

CSV 是「逗号分隔值」，医院/科研数据最常见的格式。用 csv 模块读写，比手写逗号更规范。

````python
import csv

rows = [
    ["姓名", "年龄", "诊断"],
    ["王小明", 45, "2型糖尿病"],
    ["李小红", 32, "社区获得性肺炎"],
]

with open("patients.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(rows)       # 一次写多行

print("已写入 patients.csv")
````

````python
with open("patients.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)               # 每行是一个列表
````

**练习 5（今天的检查点·CSV 读写）**：用 csv 模块把一个「患者表」写入 `patients2.csv`，再读回来逐行打印。数据：

````python
# 在这里写你的代码
import csv

rows = [
    ["姓名", "年龄", "WBC"],
    ["王小明", 45, 4.2],
    ["李小红", 32, 12.1],
    ["张三", 28, 2.8],
]

# 第一段：写
with open("patients2.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(rows)

# 第二段：读
with open("patients2.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
````
