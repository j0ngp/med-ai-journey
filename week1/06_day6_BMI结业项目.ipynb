## 📋 项目需求（像真实产品一样写清楚）

**用户**：社区医生 / 健康管理师

**程序要做的**：

1. 询问用户的体重（kg）和身高（m）；
2. 输入不是数字时不崩溃，友好提示并让用户重新输入；
3. 计算 BMI = 体重 / 身高²，保留 1 位小数；
4. 按标准分级：`<18.5` 偏瘦 ｜ `18.5~23.9` 正常 ｜ `24~27.9` 超重 ｜ `≥28` 肥胖；
5. 输出一句话报告，如：`BMI 27.7，超重，建议：控制饮食、增加运动`；
6. 循环：可以连续评估多位用户，输入 `q` 退出。

**验收输出示例**：

```
欢迎使用 BMI 健康评估（输入 q 退出）
请输入体重(kg)：80
请输入身高(m)：1.7
BMI 27.7，超重，建议：控制饮食、增加运动
请输入体重(kg)：q
再见！
```

## 🧩 分 5 步搭建（每步先自己写，卡住再往下看提示）

### 第 1 步：安全的输入函数

写 `safe_input(提示文字)`：循环询问，直到用户输入合法数字才返回 float；输入 `q` 则返回 `"q"`。

提示：while + try/except ValueError + break。

````python
# 在这里写你的代码
def safe_input(prompt):
    while True:                          # 一直问
        text = input(prompt)             # 先当字符串拿
        if text == "q":                  # 想退出 → 返回 "q"
            return "q"
        try:
            return float(text)           # 能转小数 → 返回数字
        except ValueError:               # 转不了 → 提示后回到循环开头
            print("输入无效，请输入数字（或输入 q 退出）")
````

````python
safe_input("请输入体重(kg)：")
````

### 第 2 步：BMI 计算函数

写 `bmi_calc(weight_kg, height_m)`，返回保留 1 位小数的 BMI。

````python
# 在这里写你的代码
def bmi_calc(weight_kg, height_m):
    return round(weight_kg/height_m**2,1)
````

````python
print(bmi_calc(80, 1.7))     # 预期 27.7
print(bmi_calc(60, 1.65))    # 预期 22.0
````

### 第 3 步：分级 + 建议函数

写 `interpret_bmi(bmi)`，返回 `(等级, 建议)` 两个值：

| BMI | 等级 | 建议 |
|---|---|---|
| <18.5 | 偏瘦 | 加强营养，规律饮食 |
| <24 | 正常 | 保持良好生活习惯 |
| <28 | 超重 | 控制饮食、增加运动 |
| ≥28 | 肥胖 | 建议就医，制定减重计划 |

提示：`return 等级, 建议`（两个值用逗号，返回一个元组，外面用 `level, advice = interpret_bmi(bmi)` 接住）。

````python
# 在这里写你的代码
def interpret_bmi(bmi):
    if bmi<18.5:
        return "偏廋","加强营养，规律饮食"
    elif bmi<24:
        return "正常","保持良好生活习惯"
    elif bmi<28:
        return "超重","控制饮食，增加运动"
    return "肥胖","建议就医，制定减重计划" 
````

````python
for b in [17.0, 22.5, 26.0, 30.0]:
    level, advice = interpret_bmi(b)     # 两个变量接住两个返回值
    print(b, "→", level, "|", advice)
````

### 第 4 步：报告函数

写 `bmi_report(weight_kg, height_m)`，内部调用第 2、3 步的函数，用 f-string 拼出：`BMI 27.7，超重，建议：控制饮食、增加运动`。

````python
# 在这里写你的代码
def bmi_report(weight_kg, height_m):
    bmi = bmi_calc(weight_kg, height_m)     # 先算，存进 bmi
    level, advice = interpret_bmi(bmi)      # 再分级，拆包接住
    return f"BMI {bmi}，{level}，建议：{advice}"
````

````python
print(bmi_report(80, 1.7))    # BMI 27.7，超重，建议：控制饮食，增加运动
print(bmi_report(60, 1.65))   # BMI 22.0，正常，建议：保持良好生活习惯
print(bmi_report(45, 1.7))    # BMI 15.6，偏瘦，建议：加强营养，规律饮食
````

### 第 5 步：主程序循环

把第 1、4 步接起来：欢迎语 → 循环询问体重、身高（任一为 `q` 就退出）→ 打印报告 → 结束语。

````python
# 在这里写你的代码
print("欢迎使用 BMI 健康评估（输入 q 退出）")

while True:
    weight = safe_input("请输入体重(kg)：")
    if weight == "q":
        break
    height = safe_input("请输入身高(m)：")
    if height == "q":
        break
    print(bmi_report(weight, height))

print("再见！")
````

### 第 6 步（加分）：批量报告

给一组患者批量评估：

```python
patients = [(80, 1.7), (45, 1.65), (90, 1.75)]
```

用 `for w, h in patients:` 逐个打印报告。

````python
# 在这里写你的代码
patients = [(80, 1.7), (45, 1.65), (90, 1.75)]

for w, h in patients:
    print(bmi_report(w, h))
````
