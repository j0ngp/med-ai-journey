## 任务二 · 「模型输出 text → 代码定位下标」

评分卡「起止下标」，但模型算不准字符位置。正确链路是：**模型只输出 text → 你用 `str.find(text)` 定位下标**。先跑通这段代码（用 D33 的 20 条句子里随便一条试）。

```python
# 下标定位：模型输出 text，代码算出起止位置
def locate(sentence, text):
    start = sentence.find(text)          # 找到 text 第一个出现的位置（0 从 0 开始数）
    if start == -1:
        return None, None
    end = start + len(text)              # 结束 = 起始 + 字数
    return start, end

s = "太阳病，头痛发热，汗出恶风，桂枝汤主之。"
for t in ["头痛", "桂枝汤", "太阳病"]:
    start, end = locate(s, t)
    print(f"「{t}」 0-based半开: [{start}, {end})   1-based闭: [{start+1}, {end}]")

# 验证：切片 [start:end] 应该正好等于实体原文
start, end = locate(s, "头痛")
print("\n切片验证:", repr(s[start:end]), "== 头痛 ?", s[start:end] == "头痛")
```

## 关键提醒：下标约定有两种，等官方样例一到就对照

先记住两种主流约定（代码里都打印出来了）：
- **0-based 半开区间**：`[start, end)`，从 0 数；
- **1-based 闭区间**：`[start+1, end]`，从 1 数。

官方样例一到，你把 2–3 条「原文 + 起止位置」对照这两个方案，哪个对得上就用哪个——**这个判断我到时候帮你做**，今天先把 `str.find` 这条链路练熟。
