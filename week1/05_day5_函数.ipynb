## 1. 什么是函数：把一段逻辑打包成「机器」

函数 = 一个处理数据的机器：放进原料（参数）→ 吐出成品（返回值）。

```
def 函数名(参数):
    处理逻辑
    return 结果
```

````python
def greet(name):
    return "你好，" + name + "！"

print(greet("王小明"))
print(greet("李小红"))
````

### 1.1 return 和 print 的区别（今天最重要的概念）

- `print()`：把内容**显示到屏幕上**，显示完就没了；
- `return`：把结果**交还给调用者**，可以存进变量、继续参与计算。

````python
def double(x):
    return x * 2        # 把结果还回去

result = double(5)       # result 接住了返回值
print(result)            # 10
print(double(7) + 3)     # 17，返回值能直接参与运算
````

## 2. 多参数 + 默认参数

````python
def bmi_calc(weight_kg, height_m):
    """计算 BMI：体重(kg) / 身高(m)的平方"""
    return weight_kg / (height_m ** 2)

print(bmi_calc(70, 1.75))       # 22.86
print(round(bmi_calc(70, 1.75), 1))   # round 保留 1 位小数 → 22.9
````

````python
def judge_temp(temp, fever_line=37.3):   # 默认参数：不传就用 37.3
    if temp >= fever_line:
        return "发热"
    return "正常"

print(judge_temp(38.5))        # 发热（用默认 37.3）
print(judge_temp(36.5))        # 正常
print(judge_temp(38.0, 38.5))  # 38.5 这条线以上才发热 → 正常
````

## 3. 函数 + 循环：批量处理（以后做评测就是这招）

````python
def classify_wbc(wbc):
    if wbc < 3.5:
        return "偏低"
    elif wbc <= 9.5:
        return "正常"
    return "偏高"

wbc_list = [4.2, 12.1, 2.8, 7.5]

for w in wbc_list:
    print("WBC", w, "→", classify_wbc(w))
````

## 4. try / except —— 出错也不崩（今天看懂即可）

用户输入了「不是数字」的东西，`int()` 会报错。用 try/except 兜住：

````python
def safe_int(text):
    try:
        return int(text)        # 先试着转
    except ValueError:          # 转不了（ValueError）就执行这里
        return None             # None 表示「没有结果」

print(safe_int("123"))     # 123
print(safe_int("abc"))     # None（不会崩溃）
````

```python
patients = [(60, 1.65), (80, 1.70), (45, 1.75)]   # (体重kg, 身高m)
```

用 for 循环逐个调用并打印。

````python
# 在这里写你的代码
def bmi_report(weight_kg, height_m):
    bmi = weight_kg / height_m ** 2
    return f"BMI {round(bmi, 1)}，{interpret_bmi(bmi)}"

patients = [(60, 1.65), (80, 1.70), (45, 1.75)]
for w, h in patients:
    print(bmi_report(w, h))
````

**练习 6（加分题）**：写函数 `safe_float(text)`，用 try/except 把字符串转小数，转不了返回 `None`。测试 `"36.5"` 和 `"发烧"`。

````python
# 在这里写你的代码
def safe_float(text):
    try:
        return float(text)
    except ValueError:
        return None
print(safe_float("36.5"))
print(safe_float("发烧"))
````

定义：`def 函数名(参数):` → 逻辑 → `return 结果`
