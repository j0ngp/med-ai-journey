## 1. 什么是 pandas / DataFrame

pandas 是 Python 处理**表格数据**的库。它最核心的东西叫 **DataFrame**——可以理解为「Excel 表格」：有行、有列、每列有名字。

惯例：`import pandas as pd`（借进来，简称 pd）。

```python
import pandas as pd

df = pd.read_csv("patients_demo.csv")   # 一行读入 CSV，得到 DataFrame

print(type(df))     # <class 'pandas.core.frame.DataFrame'>
df                   # 直接显示整个表格
```

## 2. 查看数据的 5 个必学方法

```python
df.head()        # 前 5 行（括号里可写数字，如 head(3)）
```

```python
print(df.shape)       # (行数, 列数)
print(df.columns)     # 所有列名
```

```python
print(df.dtypes)      # 每列的数据类型（object=字符串，int64=整数，float64=小数）
```

```python
df.info()             # 概览：行列数、每列类型、有无缺失值
```

## 3. 取一列（得到 Series）

`df["列名"]` 取出一列，变成一个 Series（可以理解成「只有一列的表格」）。

```python
ages = df["年龄"]
print(ages)                 # 一整列
print(type(ages))           # Series
```

### 3.1 对一列做统计（今天核心）

```python
print("平均年龄:", df["年龄"].mean())       # 均值
print("最大年龄:", df["年龄"].max())
print("最小年龄:", df["年龄"].min())
print("中位年龄:", df["年龄"].median())
print("血糖平均值:", df["血糖"].mean())
```

```python
df.describe()       # 一次性看所有数值列的统计（count/mean/std/min/四分位/max）
```

## 4. 计数：value_counts（医学数据超常用）

统计一列里每个值出现了多少次——比如「每种诊断各有多少人」。

```python
df["诊断"].value_counts()      # 每种诊断的人数
```

```python
df["性别"].value_counts()      # 男女各多少人
```
