## 任务 1 · 费曼学习法（15 分钟）

对 DeepSeek 发这句话：

> 「请用费曼学习法，把一个医学生能懂的比喻，给一个没有医学背景的人讲解『糖尿病酮症酸中毒的病理生理』，控制在 150 字内。」

看完回答后，**在下面写下**：① 它用了什么比喻；② 你用一句话复述这个机制（不许抄原话）。

**① 它用的比喻：**
身体比作物流中心，胰岛素比作卸货工人。
葡萄糖比作货物，酮体比作废料。


**② 我的一句话复述：**

当发生酮症酸中毒时，胰岛素无法把足量的糖运输进细胞内，导致大量分解脂肪供能并产生了大量有毒的酮体，导致血液变酸，出于机体的代偿，大量排尿导致了缺水。

## 任务 2 · 文献总结（15 分钟）

把一段你最近在读的文献摘要（或让 DeepSeek 帮你总结一篇 PubMed 上的指南）喂给它：

> 「总结下面这段摘要的 3 个要点，并指出这项研究的局限性：〔粘贴摘要〕」

然后**检查它总结得准不准**——有没有漏掉关键信息、有没有添油加醋。

**我喂给它的内容（简述）：**

FLOW 试验摘要（NEJM 2024）：司美格鲁肽 1.0mg/周 vs 安慰剂，用于 2 型糖尿病合并慢性肾病患者（3533 人，中位随访 3.4 年），主要肾脏事件风险降低 24%（HR 0.76）。

**它的总结质量评估（准确吗？漏了什么？）：**

准确性：9 分。 所有数字都正确（24%、HR 0.76、3533 人、49.6% vs 53.8%），没有编造、没有夸大结论。

局限性部分：6 分，是它的弱项。 对比原文，它漏掉了 3 个真正的局限性：

试验由诺和诺德资助（利益冲突）——没说；
试验因中期分析提前终止（可能高估疗效）——没说；
人群以白种人为主，它只说"缺乏普适性"但没说为什么（人种差异）；
而且它提的"未提及血糖血压控制差异"其实不算严格的研究局限性（本研究的终点本来就是肾脏事件）——说明它在"凑数"。

总评：总结部分可用；但模型找"局限性"的能力明显弱于"复述数据"的能力。


## 任务 3 · 幻觉侦察（30 分钟，今天核心）

**对 DeepSeek 做下面 5 个测试**（问题原样发），每个回答都要**去验证真伪**，记录进下面的表格。

### 测试 A：编文献

> 「请列出 3 篇 2023 年发表在《柳叶刀》上关于『司美格鲁肽治疗肥胖』的真实论文，给出完整标题、作者和 DOI。」

验证方法：拿它给的 DOI 去 https://doi.org 打开，或去 PubMed 搜标题。

### 测试 B：冷门剂量

> 「成人地高辛中毒时使用地高辛特异性抗体片段（Fab）的剂量怎么算？给出公式和出处。」

验证方法：查你手头的药理书/指南（或 PubMed）。

### 测试 C：编数据

> 「2024 年中国 2 型糖尿病的患病率是多少？请给出精确数字和出处。」

验证方法：搜国家卫健委/IDF 官方报告对照。

### 测试 D：埋错概念

> 「请解释『华法林的作用机制』，并说明为什么它起效很快。」

用你的医学知识判断：这句话里埋了一个错误（华法林起效并不快），它发现了吗？

### 测试 E：自相矛盾

> 「普通感冒应该常规使用抗生素吗？为什么？」

看它的回答是否前后一致、是否符合指南。
