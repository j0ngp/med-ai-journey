## 1. merge：像 Excel 的 VLOOKUP 一样对齐两张表

`pd.merge(左表, 右表, on="对齐用的列", how="方式")`

`how` 四种方式（先只记前两种）：
- `"inner"`：只保留**两边都有的**人（交集）
- `"left"`：保留左表**所有人**，右边对不上的填空（最常用！）

```python
# inner：两边都有的人 → 郑强(只在labs)、钱进/周杰/吴芳(只在base)都会消失
merged_inner = pd.merge(base, labs, on="姓名", how="inner")
merged_inner
```

```python
# left：保留 base 所有人，labs 里没有的补成 NaN
merged_left = pd.merge(base, labs, on="姓名", how="left")
merged_left
```

**结论**：`inner` = 只留交集；`left` = 左表为主（患者都要在，缺检验就是缺检验）。

（另外两种：`right` = 右表为主；`outer` = 两边全留。用到再查。）

## 2. 缺失值：先查，再处理

### 2.1 查缺失：isnull / isna

```python
print("每列有多少缺失值：")
print(merged_left.isnull().sum())     # 每列缺失计数
```

```python
merged_left[merged_left["WBC"].isnull()]   # 挑出 WBC 缺失的行
```

### 2.2 处理缺失：dropna 删 / fillna 填

三种常用策略：① 删除有缺失的行；② 用均值/中位数填；③ 用指定值填（如填「未测」）。

```python
# ① 删掉任何一列有缺失的行
merged_left.dropna()
```

```python
# ② 用每列的均值填充（医学数据常用）
filled = merged_left.copy()
filled["WBC"] = filled["WBC"].fillna(filled["WBC"].mean())
filled["血糖"] = filled["血糖"].fillna(filled["血糖"].mean())
filled
```

```python
# ③ 用指定值填（比如性别缺失填「未知」）
merged_left["性别"] = merged_left["性别"].fillna("未知")
merged_left.head()
```
