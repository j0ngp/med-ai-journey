## 2. 变量 —— 给数据起名字

变量就是「贴了标签的盒子」。医学场景里我们可以这样存患者信息：

```python
patient_name = "王小明"    # 字符串要用引号
age = 45                   # 整数不用引号
temperature = 38.5         # 小数
is_fever = True            # 布尔：是/否

print(patient_name, "年龄", age, "体温", temperature)
```

**变量命名规则（今天先记住两条）**：
1. 用英文字母、数字、下划线，不能以数字开头（`age1` ✅，`1age` ❌）；
2. 见名知义：`age` 比 `a` 好，`patient_name` 比 `pn` 好。

## 3. 四种基本数据类型

```python
print(type(45))        # int   整数
print(type(38.5))      # float 小数（浮点数）
print(type("发烧"))     # str   字符串
print(type(True))      # bool  布尔（True / False）
```

**type() 可以查看任何数据的类型**——以后程序报错，先 type() 一下往往就明白了。

## 4. 类型转换（医学场景常用）

从外部拿到的数据经常是「长得像数字的字符串」，做计算前要先转换：

```python
age_str = "45"           # 这是字符串
age = int(age_str)       # 转成整数
print(age + 1)           # 46

print(float("38.5"))     # 字符串转小数
print(str(38.5))         # 数字转字符串
print(int(38.9))         # 小数转整数会直接去掉小数部分（不是四舍五入）
```

## 5. 数字运算

```python
print(10 + 3)    # 加
print(10 - 3)    # 减
print(10 * 3)    # 乘
print(10 / 3)    # 除（结果是小数）
print(10 // 3)   # 整除（只要商，去掉小数）
print(10 % 3)    # 取余（余数）
print(2 ** 10)   # 幂：2 的 10 次方
```

## 6. input —— 让程序提问

```python
name = input("你叫什么名字？")
print("你好，", name, "！")
```

⚠️ **input 拿到的永远是字符串**。如果输入的是数字，要先用 int() 或 float() 转换：

```python
age = int(input("你今年几岁？"))
print("明年你", age + 1, "岁")
```
