## 1. import 的三种写法

`import` = 把 Python 自带的工具箱「借进来」。三种写法：

````python
import os                    # 方式 1：借整个模块，用 os.xxx 调用
print(os.name)

from datetime import date    # 方式 2：只借一个功能，直接用
print(date.today())

import json as js            # 方式 3：借进来改个短名
print(js.dumps({"a": 1}))
````

## 4. random 模块：随机（做抽样/分组常用）

````python
import random

patients = ["王小明", "李小红", "张三", "赵四"]
print("随机抽 1 个:", random.choice(patients))
print("随机抽 2 个:", random.sample(patients, 2))
print("随机整数 1-10:", random.randint(1, 10))
````

## 5. json 模块（今天最重要）

JSON 是一种「通用数据格式」，长得和 Python 字典几乎一样。**存结构化数据、调 API、存评测结果都用它**。

两个方向记牢：
- `dumps`（有 s）= 字典 → **字符串**（string）
- `loads`（有 s）= 字符串 → **字典**
- `dump`（无 s）= 字典 → **写进文件**
- `load`（无 s）= 从**文件读出** → 字典

````python
import json

patient = {"姓名": "王小明", "年龄": 45, "WBC": 4.2}

text = json.dumps(patient, ensure_ascii=False)   # 字典 → 字符串（中文不转义）
print("转成字符串:", text)
print("类型:", type(text))

back = json.loads(text)                          # 字符串 → 字典
print("转回字典:", back)
print("类型:", type(back))
````

````python
# 存成 json 文件，再读回来
with open("patient.json", "w", encoding="utf-8") as f:
    json.dump(patient, f, ensure_ascii=False)    # dump 写文件（无 s）

with open("patient.json", "r", encoding="utf-8") as f:
    data = json.load(f)                          # load 读文件（无 s）

print("从文件读回:", data)
print(data["姓名"])
````
