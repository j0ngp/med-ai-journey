## 1. 列表 list —— 一组有序数据（用方括号 `[]`）

医学场景：一组患者名单、一组检验项目、一组血糖测量值。

````python
patients = ["王小明", "李小红", "张三"]   # 字符串列表
glucose = [5.2, 6.1, 4.8, 7.0, 5.5]        # 小数列表

print(patients)
print(glucose)
print(len(patients))    # len() 数一数有多少个元素
````

### 1.1 取值：用「下标」从 0 开始数

`patients[0]` 是第 1 个，`patients[2]` 是第 3 个，`patients[-1]` 是倒数第 1 个。

````python
patients = ["王小明", "李小红", "张三"]

print(patients[0])     # 第 1 个
print(patients[2])     # 第 3 个
print(patients[-1])    # 最后一个
print(patients[0:2])   # 切片：第 1、2 个（不含下标 2）
````

### 1.2 增、删、改

````python
labs = ["WBC", "HGB", "PLT"]   # 检验项目

labs.append("CRP")      # 末尾追加
labs.insert(1, "NEUT")  # 在下标 1 的位置插入
print("追加后:", labs)

labs[0] = "wbc"          # 修改第 1 个元素
labs.remove("PLT")       # 按内容删除
print("修改删除后:", labs)

print("CRP 在列表里吗？", "CRP" in labs)   # in 判断是否存在
````

## 2. 字典 dict —— 带标签的数据（用花括号 `{}`）

每个数据配一个「键（key）」，像病历卡：姓名、年龄、血压……
格式：`{键: 值, 键: 值}`

````python
patient = {
    "name": "王小明",   # 键是字符串，要加引号
    "age": 45,
    "sbp": 110,          # 收缩压
    "temp": 38.5
}

print(patient)
print(patient["name"])      # 用键取值
print(patient["temp"])
````

### 2.1 用 get() 安全取值（推荐习惯）

用 `patient["xxx"]` 取一个不存在的键会**报错**；用 `get()` 可以给默认值：

````python
print(patient.get("dbp", "没有记录"))   # dbp 不存在，返回默认值
print(patient.get("name", "没有记录"))  # 存在，正常返回
````

### 2.2 增、改、删、遍历

````python
patient["hospitalized"] = True   # 加一对新键值
patient["temp"] = 37.2           # 改已有的键的值
print(patient)

print("所有键:", list(patient.keys()))
print("所有值:", list(patient.values()))

patient.pop("sbp")               # 删掉一对键值
print("删除 sbp 后:", patient)
````

## 3. 元组 tuple —— 不能修改的列表（用圆括号 `()`）

适合存「固定的参考数据」，比如正常参考范围。一旦创建就不能增删改。

````python
normal_range = (3.5, 5.5)     # 空腹血糖正常范围 mmol/L

print(normal_range[0])        # 3.5
print(6.1, "在范围内吗？", 3.5 <= 6.1 <= 5.5)
print(4.8, "在范围内吗？", 4.8 in normal_range)   # in 也能用于元组

#normal_range[0] = 3.6  # 取消这行注释运行试试 → 会报错！这就是元组的「不可修改」
````
