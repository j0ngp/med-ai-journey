```python
import pandas as pd
df = pd.read_csv("patients_demo.csv")
df.head()
```

## 1. 布尔筛选：df[条件]

核心思路：写一个「条件」，它对每一行判断 True/False，df[条件] 就只保留 True 的行。

先看条件长什么样：

```python
df["年龄"] > 60        # 每一行判断：年龄是否大于 60 → True/False 一串
```

```python
# 把条件套进方括号，只保留 True 的行
df[df["年龄"] > 60]
```

### 1.1 等于：用两个等号 ==

挑出所有「糖尿病」患者：

```python
df[df["诊断"] == "糖尿病"]
```

### 1.2 不等于：!=（挑出所有「不是健康」的人）

```python
df[df["诊断"] != "健康"]
```

## 2. 多个条件：&（且）和 |（或）

⚠️ 两个铁律：
1. 用 `&`（且）、`|`（或），**不是** `and`/`or`；
2. **每个条件都要加括号**。

```python
# 男性 且 糖尿病（两个条件都满足）
df[(df["性别"] == "男") & (df["诊断"] == "糖尿病")]
```

```python
# 年龄 > 60 或 收缩压 > 150（满足其一即可）
df[(df["年龄"] > 60) | (df["收缩压"] > 150)]
```

## 3. isin：属于列表里的值

挑出诊断是「糖尿病」或「高血压」的人（比写两个 == 再 | 更简洁）：

```python
df[df["诊断"].isin(["糖尿病", "高血压"])]
```

## 4. 排序：sort_values

```python
df.sort_values("收缩压", ascending=False)   # 按收缩压从高到低（False=降序）
```

```python
df.sort_values("年龄")     # 默认升序（从小到大）
```
