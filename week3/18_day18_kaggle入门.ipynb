## 第 3 步 · 第一个「模型」：用常识预测

历史上泰坦尼克是「妇女儿童优先」逃生。所以最朴素的预测规则：**女性→生还，男性→遇难**。

用 pandas 写这个规则（这是你全部需要的新知识）：

```python
test["Survived"] = (test["Sex"] == "female").astype(int)
```

读法：`test["Sex"] == "female"` 得到 True/False 一串；`.astype(int)` 把 True/False 变成 1/0。

### 练习 3：读入测试集，用性别规则生成预测列

````python
# 在这里写你的代码
import pandas as pd
df=pd.read_csv("titanic_test.csv")
df["Survived"] = (df["Sex"] == "female").astype(int)
df.head()
````

## 第 5 步 · 检验自己预测得准不准（本地评分）
